# Arithmetic

```python
import numpy as np
```

### Scalar

```python
array = np.array([1, 2, 3, 4, 5])

print("Original array:", array)
print("Array after adding 10:", array + 10)
print("Array after subtracting 5:", array - 5)
print("Array after multiplying by 2:", array * 2)
print("Array after dividing by 2:", array / 2)
```

### Vectorized Math Functions

```python
import numpy as np
```

```python
array = np.array([1, 2, 3, 4, 5])

print(np.sqrt(array)) # Square root
print(np.round(array / 3, 2)) # Rounding to 2 decimal 
print(np.exp(array)) # Exponential
print(np.dot(array, [2, 2, 2, 2, 2])) # Dot product
```
